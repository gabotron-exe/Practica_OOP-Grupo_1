```python
class User:
    
    def __init__(self, nombre: str, saldo: int|float, checking_account:bool = False):
        if saldo < 0:
            raise ValueError("No se puede crear un usuario con saldo negativo.")
        self.nombre = str(nombre)
        self.saldo = int(saldo)
        self.checking_account = bool(checking_account)
        
    def withdraw (self, cantidad: int):
        self.cantidad = cantidad
        if cantidad < 0:
            raise ValueError
        
        elif not isinstance (cantidad, int|float):
            raise TypeError ("Todas las cantidades han de ser números enteros o con decimales.")
        
        elif cantidad > self.saldo:
            raise ValueError ("No hay suficiente saldo para esta operación.")
        self.saldo -= cantidad
        self.saldo = round(self.saldo)
        
        return f"Cuenta a nombre de {self.nombre} tiene un saldo disponible de {self.saldo}."
    
    
    def deposit (self, cantidad: int|float):
        self.cantidad = cantidad
        
        if cantidad < 0:
            raise ValueError ("Los depósitos deben ser superiores a 0")
        elif not isinstance (cantidad, int|float):
            raise TypeError ("Todas las cantidades han de ser números enteros o con decimales.")
        self.saldo += cantidad
        self.saldo = round(self.saldo)
        return (f"Cuenta a nombre de {self.nombre} tiene un saldo disponible de {self.saldo}.")
        
        
        
    def transfer (self, cantidad: int|float, otro_usuario: "user"):
        self.cantidad = cantidad
        
        if not isinstance (cantidad, int|float):
            raise TypeError ("Todas las cantidades han de ser números enteros o con decimales.")
        
        if cantidad < 0:
            raise ValueError
        
        elif otro_usuario.checking_account == False:
            raise ValueError ("La cuenta de destino no existe")
        self.otro_usuario = otro_usuario
        
        if cantidad > self.saldo:
            raise ValueError ("No hay suficiente saldo para esta operación.")
        
        self.saldo -= cantidad
        self.saldo = round(self.saldo)
        otro_usuario.saldo = otro_usuario.saldo + cantidad
        otro_usuario.saldo = round(otro_usuario.saldo)
        
        return (f"Se han transferido {cantidad} a la cuenta de {otro_usuario}. {self.usuario} tiene un saldo restante de {self.saldo}. {otro_usuario} tiene un saldo restante de{otro_usuario.saldo}")
```

```python
# Pruebas de aceptación. No hace falta que toquéis esta celda: ejecutadla para ver cómo vais.
# Una empresa usa pruebas así para saber que el código hace lo acordado.

resultados = []


def prueba(descripcion, esperado, funcion, *args):
    """Ejecuta funcion(*args) y compara el resultado con lo esperado.

    esperado puede ser un valor o un tipo de error, como ValueError.
    """
    try:
        obtenido = funcion(*args)
    except Exception as error:
        obtenido = type(error)
    correcto = obtenido == esperado
    resultados.append(correcto)
    if correcto:
        print(f"OK     {descripcion}")
    else:
        print(f"FALLA  {descripcion}\n         esperado: {esperado}\n         obtenido: {obtenido}")


def tipo_de(objeto, atributo):
    return type(getattr(objeto, atributo))


def tarea_1():
    print("--- Tarea 1: constructor y validación ---")
    jeff = User("Jeff", 70, True)
    prueba("guarda el nombre", "Jeff", getattr, jeff, "nombre")
    prueba("guarda el saldo", 70, getattr, jeff, "saldo")
    prueba("guarda checking_account", True, getattr, jeff, "checking_account")
    prueba("checking_account es bool", bool, tipo_de, User("Ana", 1, 1), "checking_account")
    prueba("saldo float: quita los decimales", 10, getattr, User("Ana", 10.9, True), "saldo")
    prueba("saldo float: queda como int", int, tipo_de, User("Ana", 10.0, True), "saldo")
    prueba("saldo negativo: ValueError", ValueError, User, "Ana", -1, True)
    prueba("saldo de texto: TypeError", TypeError, User, "Ana", "mil", True)


def tarea_2():
    print("--- Tarea 2: deposit y withdraw ---")
    ana = User("Ana", 100, True)
    prueba("deposit devuelve el mensaje", "Ana tiene un saldo de 120.", ana.deposit, 20)
    prueba("withdraw devuelve el mensaje", "Ana tiene un saldo de 118.", ana.withdraw, 2.9)
    prueba("withdraw sin saldo: ValueError", ValueError, ana.withdraw, 1000)
    prueba("deposit negativo: ValueError", ValueError, ana.deposit, -5)
    prueba("withdraw negativo: ValueError", ValueError, ana.withdraw, -5)
    prueba("deposit de texto: TypeError", TypeError, ana.deposit, "diez")
    prueba("los errores no cambian el saldo", 118, getattr, ana, "saldo")


def tarea_3():
    print("--- Tarea 3: los ejemplos del enunciado ---")
    jeff = User("Jeff", 70, True)
    joe = User("Joe", 70, False)
    prueba("withdraw(2)", "Jeff tiene un saldo de 68.", jeff.withdraw, 2)
    prueba("Joe transfiere 50", "Joe tiene un saldo de 20 y Jeff tiene un saldo de 118.", joe.transfer, jeff, 50)
    prueba("Joe no puede recibir: ValueError", ValueError, jeff.transfer, joe, 80)
    prueba("ese error no cambia a Jeff", 118, getattr, jeff, "saldo")
    prueba("ese error no cambia a Joe", 20, getattr, joe, "saldo")
    joe.checking_account = True
    prueba("Jeff transfiere 80", "Jeff tiene un saldo de 38 y Joe tiene un saldo de 100.", jeff.transfer, joe, 80)
    prueba("Jeff no tiene 100: ValueError", ValueError, jeff.transfer, joe, 100)
    prueba("transferencia negativa: ValueError", ValueError, jeff.transfer, joe, -1)
    prueba("deposit(20.00)", "Jeff tiene un saldo de 58.", jeff.deposit, 20.00)


def ejecutar_pruebas():
    resultados.clear()
    for tarea in (tarea_1, tarea_2, tarea_3):
        try:
            tarea()
        except Exception as error:
            print(f"PENDIENTE  {type(error).__name__}: {error}")
        print()
    print(f"Resultado: {sum(resultados)} de {len(resultados)} pruebas correctas.")


ejecutar_pruebas()
```
